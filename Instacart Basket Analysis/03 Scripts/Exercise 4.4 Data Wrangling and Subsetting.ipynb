## 10 You also need to provide some details about this user’s behavior. What basic stats can you provide based on the information you have?

```python
### 1. Total number of orders placed by the user
total_orders = df_user_1['order_number'].nunique()

### 2. Most frequent order hour
most_frequent_hour = df_user_1['order_hour_of_day'].mode()[0]

### 3. Most frequent day of the week for orders
most_frequent_day = df_user_1['orders_day_of_week'].mode()[0]

### 4. Average days since prior order
average_days_between_orders = df_user_1['days_since_prior_order'].mean()

### 5. Min and max days since prior order (to identify any unusually long or short gaps)
min_days_between_orders = df_user_1['days_since_prior_order'].min()
max_days_between_orders = df_user_1['days_since_prior_order'].max()

### 6. Display the basic statistics
print(f"User 1 placed {total_orders} orders.")
print(f"The most frequent order hour is {most_frequent_hour}.")
print(f"The most frequent day of the week for orders is {most_frequent_day}.")
print(f"The average days between orders is {average_days_between_orders:.2f} days.")
print(f"The minimum days between orders is {min_days_between_orders} days.")
print(f"The maximum days between orders is {max_days_between_orders} days.")
```
