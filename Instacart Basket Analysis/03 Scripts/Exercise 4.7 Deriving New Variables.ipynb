# 4 When too many users make Instacart orders at the same time, the app freezes. 
The senior technical officer at Instacart wants you to identify the busiest hours of the day. Rather than by hour, they want periods of time labeled “Most orders,” “Average orders,” and “Fewest orders.” Create a new column containing these labels called “busiest_period_of_day.”

```python
# Trying to define what qualifies as "Most orders," "Average orders," and "Fewest orders."
ords_prods_merge['order_hour_of_day'].describe()
```

# Defining the categories based on the quartiles
Most Orders: Hours around the 50% and 75% range (1 PM to 4 PM) might represent the busiest times.
Average Orders: Hours near the median but outside the busiest 25%.
Fewest Orders: The lowest quartile hours (early morning and late night).

```python
# "Most orders" for the busiest hours (1 PM to 4 PM)
ords_prods_merge.loc[(ords_prods_merge['order_hour_of_day'] >= 13) & 
                     (ords_prods_merge['order_hour_of_day'] <= 16), 'busiest_period_of_day'] = 'Most orders'
```

```python
# "Average orders" for moderately busy hours (10 AM to 12 PM and 5 PM to 8 PM)
ords_prods_merge.loc[((ords_prods_merge['order_hour_of_day'] >= 10) & 
                      (ords_prods_merge['order_hour_of_day'] <= 12)) |
                     ((ords_prods_merge['order_hour_of_day'] >= 17) & 
                      (ords_prods_merge['order_hour_of_day'] <= 20)), 'busiest_period_of_day'] = 'Average orders'
```

```python
# "Fewest orders" for the least busy hours (midnight to 9 AM and 9 PM to 11 PM)
ords_prods_merge.loc[((ords_prods_merge['order_hour_of_day'] >= 0) & 
                      (ords_prods_merge['order_hour_of_day'] <= 9)) |
                     ((ords_prods_merge['order_hour_of_day'] >= 21) & 
                      (ords_prods_merge['order_hour_of_day'] <= 23)), 'busiest_period_of_day'] = 'Fewest orders'
```

```python
# 5 Print the frequency for this new column
ords_prods_merge['busiest_period_of_day'].value_counts(dropna = False)
```
