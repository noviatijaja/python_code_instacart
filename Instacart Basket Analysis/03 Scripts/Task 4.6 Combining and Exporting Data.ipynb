```python
# Import libraries
import pandas as pd
import numpy as np
import os
```

```python
# Define the path to the pickle file
path = r'C:\Users\User\Documents\Career Foundry\Data Analytics Immersion\Data Immersion 4. Python Fundamentals for Data Analysts\Instacart Basket Analysis'
file_path = os.path.join(path, '02 Data', 'Prepared Data', 'orders_products_combined.pkl')
```

```python
# 3 Import the orders_products_combined dataframe from the pickle file
orders_products_combined = pd.read_pickle(file_path)
```

```python
# Verify
orders_products_combined.head()
```

```python
# 4 Check the shape of the imported dataframe
orders_products_combined.shape
```

```python
# Import the products.csv file using the path variable and os library, and store it as df_prods
df_prods = pd.read_csv(os.path.join(path, '02 Data', 'Prepared Data', 'products_checked.csv'), index_col=False)
```

```python
df_prods.head()
```

```python
# Rename the existing '_merge' column in orders_products_combined if it exists
if '_merge' in orders_products_combined.columns:
    orders_products_combined = orders_products_combined.rename(columns={'_merge': 'existing_merge_status'})
```

```python
# 5 Merged the two data frames
ords_prods_merge = orders_products_combined.merge(df_prods, on = 'product_id', indicator = True)
```

```python
ords_prods_merge.head()
```

```python
# 6 Check the value counts of the merge flag
ords_prods_merge['_merge'].value_counts()
```

```python
# 7 Export the newly created dataframe as ords_prods_merge in a suitable format
ords_prods_merge.to_pickle(os.path.join(path, '02 Data', 'Prepared Data', 'ords_prods_merge.pkl'))
```
